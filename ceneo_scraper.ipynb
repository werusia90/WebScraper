```python
# 2. Definiujemy funkcję do wyciągania danych, aby kod był czytelny
def extract_review_data(review):
    return {
        "identyfikator": review.get('data-entry-id'),
        "autor": review.select_one('span.user-post__author-name').get_text(strip=True) if review.select_one('span.user-post__author-name') else None,
        "treść": review.select_one('div.user-post__text').get_text(strip=True) if review.select_one('div.user-post__text') else None,
        "ocena": review.select_one('span.user-post__score-count').get_text(strip=True) if review.select_one('span.user-post__score-count') else None,
        "rekomendacja": review.select_one('span.user-post__author-recomendation > em').get_text(strip=True) if review.select_one('span.user-post__author-recomendation > em') else None,
        "zalety": ", ".join([el.get_text(strip=True) for el in review.select('div.review-feature__item--positive')]),
        "wady": ", ".join([el.get_text(strip=True) for el in review.select('div.review-feature__item--negative')]),
        "przydatna": review.select_one('button.vote-yes span').get_text(strip=True) if review.select_one('button.vote-yes span') else "0",
        "nieprzydatna": review.select_one('button.vote-no span').get_text(strip=True) if review.select_one('button.vote-no span') else "0",
        "data_zamieszczenia": review.select('time')[0].get('datetime') if len(review.select('time')) > 0 else None,
        "data_zakupu": review.select('time')[1].get('datetime') if len(review.select('time')) > 1 else None,
    }
```

```python
# 3. Szukamy wszystkich opinii i tworzymy listę słowników
reviews_html = page_dom.select('div.js_product-review')
all_reviews_data = [extract_review_data(r) for r in reviews_html]

# 4. Tworzymy ramkę danych (DataFrame)
df = pd.DataFrame(all_reviews_data)

# 5. Wyświetlamy wynik
print(df.head())
```
